```python
import re

# Sample content from the legal document
text_content = """
Ενώπιον του κ. Εισαγγελέως Πλημμελειοδικών Αθηνών

ΜΗΝΥΣΗ Αλέξανδρου Λιακόπουλου του Νικολάου, Πουλοπούλου 37 Αθήνα
ΚΑΤΑ
1. Monyrok Simon του Luk, γεννηθείς 25/12/1983, στο Σουδάν και με ΑΦΜ 172863565
2. Igbinoghene Osamuyi Oris με αφμ 173718685, Αριθμό δελτίου ασύλου 587779 του 2019
3. Abdullah Ahmed Salim Abdirahman με αριθμό διαβατηρίου P00418764, υπηκόου Σομαλίας
Αξιόποινες πράξεις: φθορά ξένης ιδιοκτησίας
"""

# Function to extract key elements
def extract_legal_elements(text):
    # Extract the recipient (Ενώπιον...)
    recipient_match = re.search(r'Ενώπιον\s+του\s+(.+)', text)
    recipient = recipient_match.group(1) if recipient_match else "Not Found"
    
    # Extract the type of complaint (ΜΗΝΥΣΗ... Αξιόποινες πράξεις)
    complaint_type_match = re.search(r'ΜΗΝΥΣΗ(.+Αξιόποινες πράξεις:\s*[^.]+)', text)
    complaint_type = complaint_type_match.group(1).strip() if complaint_type_match else "Not Found"
    
    # Extract the complainant's details
    complainant_match = re.search(r'ΜΗΝΥΣΗ\s+(.+?),', text)
    complainant = complainant_match.group(1).strip() if complainant_match else "Not Found"
    
    # Extract the defendants' details
    defendants = re.findall(r'\d+\.\s+(.+)', text)
    
    return {
        "Recipient": recipient,
        "Type of Complaint": complaint_type,
        "Complainant": complainant,
        "Defendants": defendants
    }

# Apply the extraction
extracted_elements = extract_legal_elements(text_content)

# Display the extracted elements
print("Extracted Legal Elements:")
for key, value in extracted_elements.items():
    if isinstance(value, list):
        print(f"{key}:")
        for item in value:
            print(f"  - {item}")
    else:
        print(f"{key}: {value}")
```

```python
import spacy
from spacy.training import Example
from spacy.tokens import DocBin

# Function to adjust entity boundaries to match SpaCy tokens
def adjust_entity_boundaries(text, entities):
    doc = nlp.make_doc(text)
    new_entities = []
    
    for start, end, label in entities:
        # Adjust span to match token boundaries using 'expand' mode
        span = doc.char_span(start, end, label=label, alignment_mode='expand')
        if span is not None:
            new_entities.append((span.start_char, span.end_char, label))
        else:
            print(f"Skipping misaligned entity: '{text[start:end]}'")
    
    return new_entities

# Adjust the training data
adjusted_training_data = []
for text, annotations in training_data:
    adjusted_entities = adjust_entity_boundaries(text, annotations['entities'])
    adjusted_training_data.append((text, {"entities": adjusted_entities}))

# Check if the adjustments were made successfully
print("\nAdjusted Training Data:")
for text, annotations in adjusted_training_data:
    print(f"Text: {text}")
    print(f"Entities: {annotations['entities']}")
```

```python
# Function to adjust or remove overlapping entities
def resolve_overlapping_entities(text, entities):
    # Sort entities by start position
    entities_sorted = sorted(entities, key=lambda x: x[0])
    adjusted_entities = []
    last_end = -1
    
    for start, end, label in entities_sorted:
        if start < last_end:
            # There is an overlap, so we'll remove or adjust
            print(f"Overlap detected: '{text[start:end]}' with label '{label}'")
            continue  # Skip the overlapping entity
        else:
            adjusted_entities.append((start, end, label))
            last_end = end
    
    return adjusted_entities

# Apply adjustment to all training data
adjusted_training_data_no_overlap = []
for text, annotations in adjusted_training_data:
    adjusted_entities = resolve_overlapping_entities(text, annotations['entities'])
    adjusted_training_data_no_overlap.append((text, {"entities": adjusted_entities}))

# Display the adjusted training data for verification
print("\nAdjusted Training Data (No Overlaps):")
for text, annotations in adjusted_training_data_no_overlap:
    print(f"Text: {text}")
    print(f"Entities: {annotations['entities']}")
```

```python
import spacy
from spacy.tokens import DocBin
from spacy.training import Example
from spacy.util import minibatch, compounding
import re
from typing import List, Tuple, Dict, Optional, Iterator
import logging
from pathlib import Path
import json
from datetime import datetime
import numpy as np

# Set up logging with more detailed configuration
def setup_logging(log_file: Optional[str] = None):
    """Configure logging with optional file output."""
    log_format = '[%(asctime)s] [%(levelname)s] %(message)s'
    logging.basicConfig(
        level=logging.INFO,
        format=log_format,
        handlers=[
            logging.StreamHandler(),
            logging.FileHandler(log_file) if log_file else logging.NullHandler()
        ]
    )
    return logging.getLogger(__name__)

class LegalDocumentProcessor:
    """Processor for legal documents with NER capabilities."""
    
    VALID_LABELS = {
        "RECIPIENT", "COMPLAINANT", "DEFENDANT", "COMPLAINT_TYPE",
        "DATE", "ADDRESS", "LEGAL_REF", "COURT", "CASE_NUMBER"
    }

    def __init__(self, lang: str = "el", log_file: Optional[str] = "legal_processor.log"):
        """
        Initialize the Legal Document Processor.
        
        Args:
            lang (str): Language code (default: "el" for Greek)
            log_file (str, optional): Path to log file
        """
        self.logger = setup_logging(log_file)
        self.lang = lang
        self.nlp = spacy.blank(lang)
        self.setup_pipeline()
        self.training_history = []
        
    def setup_pipeline(self):
        """Configure the NLP pipeline with necessary components."""
        try:
            if "ner" not in self.nlp.pipe_names:
                self.ner = self.nlp.add_pipe("ner", last=True)
            else:
                self.ner = self.nlp.get_pipe("ner")
        except Exception as e:
            self.logger.error(f"Failed to setup pipeline: {e}")
            raise

    def validate_entity_labels(self, entities: List[Tuple[int, int, str]]) -> bool:
        """Validate that entity labels are in the allowed set."""
        return all(label in self.VALID_LABELS for _, _, label in entities)

    def prepare_training_data(self, raw_data: List[Tuple[str, Dict]], validate_labels: bool = True) -> List[Tuple[str, Dict]]:
        """
        Prepare and validate training data.
        
        Args:
            raw_data: List of tuples containing text and entity annotations
            validate_labels: Whether to validate entity labels against VALID_LABELS
            
        Returns:
            List of validated and cleaned training data
        """
        cleaned_data = []
        for text, annotations in raw_data:
            try:
                # Clean text and validate entities
                text = self._clean_text(text)
                if validate_labels and not self.validate_entity_labels(annotations["entities"]):
                    self.logger.warning(f"Invalid entity labels found in: {text[:100]}...")
                    continue
                    
                adjusted_entities = self._adjust_entity_boundaries(text, annotations["entities"])
                final_entities = self._resolve_overlapping_entities(text, adjusted_entities)
                
                if final_entities:
                    cleaned_data.append((text, {"entities": final_entities}))
                    
            except Exception as e:
                self.logger.error(f"Error processing training data: {e}")
                continue
                
        return cleaned_data

    def _clean_text(self, text: str) -> str:
        """Clean and normalize text with enhanced cleaning rules."""
        # Remove multiple spaces and normalize whitespace
        text = re.sub(r'\s+', ' ', text)
        # Remove zero-width spaces and other invisible characters
        text = "".join(char for char in text if char.isprintable())
        # Normalize Greek diacritics if needed
        # Add more specific cleaning rules as needed
        return text.strip()

    def _create_batches(self, training_data: List[Tuple[str, Dict]], batch_size: int) -> Iterator[List[Tuple[str, Dict]]]:
        """Create batches of training data."""
        for i in range(0, len(training_data), batch_size):
            yield training_data[i:i + batch_size]

    def train(self, 
              training_data: List[Tuple[str, Dict]], 
              n_epochs: int = 10, 
              dropout: float = 0.2,
              batch_size: int = 8,
              evaluation_data: Optional[List[Tuple[str, Dict]]] = None):
        """
        Train the NER model with enhanced features.
        
        Args:
            training_data: Prepared training data
            n_epochs: Number of training epochs
            dropout: Dropout rate for training
            batch_size: Size of training batches
            evaluation_data: Optional evaluation dataset
        """
        # Add labels to NER
        for _, annotations in training_data:
            for _, _, label in annotations["entities"]:
                self.ner.add_label(label)
        
        # Start training
        optimizer = self.nlp.begin_training()
        batch_sizes = compounding(4.0, batch_size, 1.001)
        
        for epoch in range(n_epochs):
            epoch_losses = []
            batches = list(self._create_batches(training_data, batch_size))
            
            for batch in batches:
                losses = {}
                for text, annotations in batch:
                    try:
                        example = Example.from_dict(self.nlp.make_doc(text), annotations)
                        self.nlp.update([example], drop=dropout, losses=losses)
                        epoch_losses.append(losses.get("ner", 0))
                    except Exception as e:
                        self.logger.error(f"Error during training: {e}")
                        continue
            
            avg_loss = np.mean(epoch_losses) if epoch_losses else 0
            self.training_history.append({
                "epoch": epoch + 1,
                "loss": avg_loss,
                "timestamp": datetime.now().isoformat()
            })
            
            self.logger.info(f"Epoch {epoch + 1}/{n_epochs} - Average Loss: {avg_loss:.4f}")
            
            # Evaluate on validation data if provided
            if evaluation_data:
                self._evaluate(evaluation_data)

    def _evaluate(self, eval_data: List[Tuple[str, Dict]]):
        """Evaluate model performance on evaluation data."""
        correct = 0
        total = 0
        
        for text, annotations in eval_data:
            doc = self.nlp(text)
            gold_entities = set((start, end, label) for start, end, label in annotations["entities"])
            pred_entities = set((ent.start_char, ent.end_char, ent.label_) for ent in doc.ents)
            
            correct += len(gold_entities.intersection(pred_entities))
            total += len(gold_entities)
        
        precision = correct / total if total > 0 else 0
        self.logger.info(f"Evaluation Precision: {precision:.4f}")

    def save_model(self, path: str):
        """Save the trained model and training history."""
        path = Path(path)
        try:
            # Save spaCy model
            self.nlp.to_disk(path)
            
            # Save training history
            history_path = path / "training_history.json"
            with open(history_path, 'w', encoding='utf-8') as f:
                json.dump(self.training_history, f, ensure_ascii=False, indent=2)
                
            self.logger.info(f"Model and training history saved successfully at: {path}")
        except Exception as e:
            self.logger.error(f"Error saving model: {e}")
            raise

    def load_model(self, path: str):
        """Load a trained model and training history."""
        path = Path(path)
        try:
            # Load spaCy model
            self.nlp = spacy.load(path)
            
            # Load training history if it exists
            history_path = path / "training_history.json"
            if history_path.exists():
                with open(history_path, 'r', encoding='utf-8') as f:
                    self.training_history = json.load(f)
                    
            self.logger.info(f"Model loaded successfully from: {path}")
        except Exception as e:
            self.logger.error(f"Error loading model: {e}")
            raise

    def get_training_summary(self) -> Dict:
        """Get a summary of the training history."""
        if not self.training_history:
            return {"status": "No training history available"}
            
        return {
            "total_epochs": len(self.training_history),
            "final_loss": self.training_history[-1]["loss"],
            "best_loss": min(h["loss"] for h in self.training_history),
            "training_duration": (
                datetime.fromisoformat(self.training_history[-1]["timestamp"]) -
                datetime.fromisoformat(self.training_history[0]["timestamp"])
            ).total_seconds() / 3600  # Hours
        }
```
